## Завдання 1

```python
import pandas as pd

df = pd.DataFrame({
    'Назва': ['Кобзар', '1984', 'Гобіт', 'Макова війна', 'Атомні звички', 'Місто'],
    'Автор': ['Тарас Шевченко', 'Джордж Орвелл', 'Джон Толкін', 'Ребекка Кван', 'Джеймс Клір', 'Валер’ян Підмогильний'],
    'Рік видання': [1840, 1949, 1937, 2018, 2018, 1928],
    'Ціна': [350, 320, 450, 600, 500, 280]
})

print('Усі книги:')
print(df.to_string(index=False))
print('\nСередня ціна:', df['Ціна'].mean())
print('\nКниги, видані після 2015 року:')
print(df[df['Рік видання'] > 2015].to_string(index=False))
print('\nСортування за ціною:')
print(df.sort_values('Ціна').to_string(index=False))
```

```python
import pandas as pd
from pathlib import Path

# Файл orders.csv має бути поруч із ноутбуком.
df = pd.read_csv('orders.csv')

print('Перші 10 рядків:')
print(df.head(10).to_string(index=False))
print('\nКількість замовлень кожного клієнта:')
print(df.groupby('Клієнт')['Номер замовлення'].count())
print('\nМаксимальна сума:', df['Сума'].max())
print('Мінімальна сума:', df['Сума'].min())
print('Загальна сума:', df['Сума'].sum())
```

## Завдання 3

```python
import pandas as pd

# Калорії та білки наведені орієнтовно на 100 г продукту.
df = pd.DataFrame({
    'Продукт': ['Яблуко', 'Банан', 'Морква', 'Картопля', 'Рис сухий', 'Гречка суха', 'Сир твердий', 'Молоко', 'Мигдаль', 'Волоський горіх'],
    'Категорія': ['Фрукти', 'Фрукти', 'Овочі', 'Овочі', 'Крупи', 'Крупи', 'Молочні', 'Молочні', 'Горіхи', 'Горіхи'],
    'Калорії': [52, 89, 41, 77, 365, 343, 350, 60, 579, 654],
    'Білки': [0.3, 1.1, 0.9, 2.0, 7.1, 13.3, 25.0, 3.2, 21.2, 15.2]
})

print('Усі продукти:')
print(df.to_string(index=False))
print('\nПродукти з калорійністю вище 300:')
print(df[df['Калорії'] > 300].to_string(index=False))
print('\nСередня кількість білків за категоріями:')
print(df.groupby('Категорія')['Білки'].mean())
print('\nСортування за калорійністю (спадання):')
print(df.sort_values('Калорії', ascending=False).to_string(index=False))
```

## Завдання 4

```python
import pandas as pd

df = pd.DataFrame({
    "Ім'я": ['Олег', 'Анна', 'Іван', 'Марія', 'Олег', 'Анна', 'Іван', 'Марія'],
    'Проект': ['Сайт', 'Сайт', 'API', 'Дизайн', 'API', 'Дизайн', 'Сайт', 'API'],
    'Години': [25, 18, 30, 22, 20, 15, 24, 12]
})

print('Вихідний DataFrame:')
print(df.to_string(index=False))

employee_hours = df.groupby("Ім'я")['Години'].sum()
project_hours = df.groupby('Проект')['Години'].sum()

print('\nГодини за співробітниками:')
print(employee_hours)
print('\nГодини за проектами:')
print(project_hours)
print('\nНайбільше годин:', employee_hours.idxmax(), '-', employee_hours.max())
```

## Завдання 5

```python
import pandas as pd

df = pd.DataFrame({
    'Фільм': ['Дюна', 'Інтерстеллар', 'Гобіт', 'Аватар'] * 3,
    'Місто': ['Львів'] * 4 + ['Київ'] * 4 + ['Одеса'] * 4,
    'Продані квитки': [150, 180, 120, 210, 240, 260, 190, 300, 130, 160, 110, 200]
})

print('Усі продажі:')
print(df.to_string(index=False))

film_sales = df.groupby('Фільм')['Продані квитки'].sum()
city_sales = df.groupby('Місто')['Продані квитки'].sum()

print('\nКвитки за фільмами:')
print(film_sales)
print('\nКвитки за містами:')
print(city_sales)
print('\nФільм з найбільшою кількістю продажів:', film_sales.idxmax(), '-', film_sales.max())
```

## Завдання 6

```python
import pandas as pd

clients = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5],
    "Ім'я": ['Олег', 'Анна', 'Іван', 'Марія', 'Петро'],
    'Місто': ['Львів', 'Київ', 'Одеса', 'Львів', 'Харків']
})

orders = pd.DataFrame({
    'ID клієнта': [1, 2, 3, 4, 5, 1, 2, 4],
    'Замовлення': ['Книга', 'Миша', 'Монітор', 'Клавіатура', 'Навушники', 'Зошит', 'Кабель', 'Флешка'],
    'Вартість': [400, 800, 5000, 1200, 1600, 100, 200, 350]
})

print('Клієнти:')
print(clients.to_string(index=False))
print('\nЗамовлення:')
print(orders.to_string(index=False))

merged = pd.merge(clients, orders, left_on='ID', right_on='ID клієнта')
print('\nОб’єднаний DataFrame:')
print(merged.to_string(index=False))

result = merged.groupby(['ID', "Ім'я", 'Місто'], as_index=False)['Вартість'].sum()
result = result.rename(columns={'Вартість': 'Загальна вартість'})

print('\nЗагальна вартість замовлень за клієнтами:')
print(result.to_string(index=False))
```
